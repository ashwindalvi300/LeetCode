Write a recursive function that takes a list containing other lists (of any depth) and returns a single “flat” list of all elements.


Given Input: nested = [1, [2, 3], [4, [5, 6]], 7]

Expected Output: Flattened: [1, 2, 3, 4, 5, 6, 7]

```python
def flatten(lst):
    result = []
    for item in lst:
        if type(item) == list:
            result = result + flatten(item)
        else: 
            result.append(item)

    return result

nested = [1,[2,3],[4,[5,6]],7]

print(flatten(nested))
```

{'1984': 'Orwell', 'Animal Farm': 'Orwell', 'Brave New World': 'Huxley'}

Write a function that generates the Power Set of a given set (a set of all possible subsets, including the empty set and the set itself).

Given Input: [1, 2, 3]

Expected Output:

[(), (1,), (2,), (3,), (1, 2), (1, 3), (2, 3), (1, 2, 3)]

```python
from itertools import combinations

def get_power_set(s):
    elements = list(s)
    power_set = []

    for r in range(len(elements) + 1):
        for combo in combinations(elements, r):
            power_set.append(combo)
    return power_set

my_set = {1,2,3}
print(f"Power Set : {get_power_set(my_set)}")
```

Create a function that asks for a user’s birthdate (YYYY-MM-DD) and calculates their exact age today in years, months, and days

Given Input: Birthdate: 1995-05-15, Today: 2026-01-02

Expected Output: Age: 30 years, 7 months, 18 days

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

def age_cal(b_day):
    try:
        birth_day = datetime.strptime(b_day, "%Y-%m-%d").date()
        t_day = datetime.now().date()

        diff = relativedelta(t_day, birth_day)

        return f"{diff.years} years, {diff.months} months, {diff.days} days"

    except ValueError:
        return "Invalid date format. Please use YYYY-MM-DD."

user_age = age_cal("1995-05-15")
print(f"Exact Age: {user_age}")
```

Write a function that calculates the exact number of days, hours, and minutes remaining until the upcoming New Year’s Day.

Given Input: Current Date: 2026-01-02

Expected Output: 363 days, 16 hours, 50 minutes until New Year!

```python
from datetime import datetime
def cal():
    today = datetime.now()
    
    target = datetime(today.year + 1,1, 1)

    diff = target - today

    days = diff.days

    hours, remainder = divmod(diff.seconds, 3600)
    minutes, _ = divmod(remainder, 60)

    return f"{days} days, {hours} hours, {minutes} minutes."

print(f'Time remaining: {cal()} until New Year!')
```

Create a function that takes a start date and an integer N, then calculates the date exactly N business days (Monday–Friday) in the future.

Given Input: Start: Friday, 2026-01-02, Days to add: 5

Expected Output:

Start: Friday, 2026-01-02

Business Target: Friday, 2026-01-09

```python
from datetime import datetime, timedelta

def add_bussiness_days(start_date, n):
    current_date = start_date
    days_add = 0
    while days_add < n:
        current_date =  current_date + timedelta(days = 1)
        if current_date.weekday() < 5:
            days_add = days_add + 1
    return current_date

start = datetime(2026,1,2)
result = add_bussiness_days(start, 5)

print(f"Start: {start.strftime('%A, %Y-%m-%d')}")
print(f"Business Target: {result.strftime('%A, %Y-%m-%d')}")
```

```python
def poweroftwo(n):
    temp = []
    for i in range(0, n+1):
        output = 2 ** i
        temp.append(output)

    return temp

print(f'The list is {poweroftwo(3)}')
```

 Write a function that identifies all duplicate elements in a list. The solution must run in linear time, meaning you should only traverse the list once.

 Given Input: numbers = [1, 2, 3, 2, 4, 5, 1, 6]

Expected Output: Duplicates found: {1, 2}

```python
def find_duplicates(nums):
    seen = set()
    duplicates = set()

    for x in nums:
        if x in seen:
            duplicates.add(x)
        else:
            seen.add(x)

    return duplicates

numbers = [1, 2, 3, 2, 4, 5, 1, 6]
print(f"Duplicates found: {find_duplicates(numbers)}")
```

Write a program that takes a string and reverses it using a for loop. While Python’s [::-1] shortcut is famous, reversing a string manually is a classic way to understand how sequences are constructed.

```python
string = "Python"
reversed_string = ""

for char in string:
    reversed_string = char  + reversed_string

print(f"Original: {string}")
print(f"Reversed: {reversed_string}")
```

Write a program that counts the total number of vowels and consonants in a given sentence, ignoring spaces and special characters.

```python
string = str(input("Enter the string : "))
vowels = 'aeiou'

v_count = 0
c_count = 0

for char in string.lower():
    if char.isalpha():
        if char in vowels:
            v_count = v_count + 1
        else:
            c_count = c_count + 1

print("String :", string)
print(f"The count of vowels is {v_count}")
print(f"The count of consonant is {c_count}")
```

Write a program to reverse a given integer number (e.g., 76542 should become 24567).

```python
num = int(input("Enter the number : "))
rev = 0
print(num)
while(num > 0):
    digit = num % 10
    rev = (rev * 10) + digit
    num = num // 10

print(rev)
```

Write a program to find the largest and smallest digit within a given integer (e.g., in 75869, the largest is 9 and the smallest is 5).

```python
num = int(input("Enter the number : "))
lis = []
print(num)

while(num > 0):
    digit = num % 10
    lis.append(digit)
    num = num // 10
largest = max(lis)
smallest = min(lis)

print("Largest digit:", largest)
print("Smallest digit:", smallest)
```

Write a program to check if a given number is a palindrome. A palindrome number is a number that remains the same when its digits are reversed (e.g., 121, 343).

```python
num = int(input("Enter the number : "))
original = num
rev = 0
print(num)
while num>0:
    digit = num % 10
    rev = (rev * 10) + digit
    num = num // 10
print(rev)
if original == rev:
    print("It is a palindrome")
else:
    print("It is not a palindrome")
```

Write a program to use a loop to find the factorial of a given number (e.g., 5!). The factorial of N is the product of all integers from 1 to N.

```python
num = int(input("Enter the number : "))
fact = 1
print(num)
if num == 0:
    print("factorial is 1")
elif num < 0:
    print("Factorial does not exist for negative numbers")
else:
    while num > 0:
        fact = fact * num
        num = num - 1
    print("factorial is ",fact)
```
